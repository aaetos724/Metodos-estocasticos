# Actividad 4.1 — Uso de árboles y estructuras de datos no lineales



## 1. El nodo del árbol

Cada nodo guarda los datos del estudiante y dos enlaces: `left` (hijo izquierdo) y `right`
(hijo derecho).

```python
import random


class Node:
    def __init__(self, matricula, nombre, carrera):
        self.matricula = matricula
        self.nombre = nombre
        self.carrera = carrera
        self.left = None
        self.right = None
```

## 2. Árbol Binario de Búsqueda

Todas las operaciones comparan por `matricula`. La inserción baja por el árbol hasta encontrar
un lugar vacío; la búsqueda es iterativa; la eliminación cubre los tres casos (hoja, un hijo y
dos hijos, reemplazando por el sucesor). Los recorridos devuelven la lista de nodos.

```python
class ArbolBST:
    def __init__(self):
        self.root = None

    def insertar(self, matricula, nombre, carrera):
        self.root = self._insertar(self.root, matricula, nombre, carrera)

    def _insertar(self, nodo, matricula, nombre, carrera):
        if nodo is None:
            return Node(matricula, nombre, carrera)
        if matricula < nodo.matricula:
            nodo.left = self._insertar(nodo.left, matricula, nombre, carrera)
        elif matricula > nodo.matricula:
            nodo.right = self._insertar(nodo.right, matricula, nombre, carrera)
        return nodo

    def buscar(self, matricula):
        nodo = self.root
        while nodo is not None and nodo.matricula != matricula:
            nodo = nodo.left if matricula < nodo.matricula else nodo.right
        return nodo

    def eliminar(self, matricula):
        self.root = self._eliminar(self.root, matricula)

    def _eliminar(self, nodo, matricula):
        if nodo is None:
            return None
        if matricula < nodo.matricula:
            nodo.left = self._eliminar(nodo.left, matricula)
        elif matricula > nodo.matricula:
            nodo.right = self._eliminar(nodo.right, matricula)
        else:
            if nodo.left is None:
                return nodo.right
            if nodo.right is None:
                return nodo.left
            sucesor = nodo.right
            while sucesor.left:
                sucesor = sucesor.left
            nodo.matricula = sucesor.matricula
            nodo.nombre = sucesor.nombre
            nodo.carrera = sucesor.carrera
            nodo.right = self._eliminar(nodo.right, sucesor.matricula)
        return nodo

    def inorden(self, nodo):
        if nodo is None:
            return []
        return self.inorden(nodo.left) + [nodo] + self.inorden(nodo.right)

    def preorden(self, nodo):
        if nodo is None:
            return []
        return [nodo] + self.preorden(nodo.left) + self.preorden(nodo.right)

    def postorden(self, nodo):
        if nodo is None:
            return []
        return self.postorden(nodo.left) + self.postorden(nodo.right) + [nodo]

    def contar(self, nodo):
        if nodo is None:
            return 0
        return 1 + self.contar(nodo.left) + self.contar(nodo.right)

    def altura(self, nodo):
        if nodo is None:
            return 0
        return 1 + max(self.altura(nodo.left), self.altura(nodo.right))
```

## 3. Funciones de apoyo

`formato` arma la línea de texto de un estudiante y `matricula_aleatoria` genera un número de
6 dígitos (1–999999) que no exista todavía en el árbol.

```python
def formato(nodo):
    return f"{nodo.matricula:06d} | {nodo.nombre:<16} | {nodo.carrera}"


def matricula_aleatoria(arbol):
    m = random.randint(1, 999999)
    while arbol.buscar(m) is not None:
        m = random.randint(1, 999999)
    return m
```

## 4. Carga de datos

```python
arbol = ArbolBST()

datos = [
    (1025,   "Ana López",     "Ingeniería"),
    (101008, "Luis García",   "Medicina"),
    (111050, "María Pérez",   "Derecho"),
    (102512, "Carlos Ruiz",   "Sistemas"),
    (103836, "Sofía Torres",  "Contabilidad"),
    (106056, "Jorge Sánchez", "Arquitectura"),
    (100221, "Elena Gómez",   "Psicología"),
]
for matricula, nombre, carrera in datos:
    arbol.insertar(matricula, nombre, carrera)

for x in arbol.inorden(arbol.root):
    print(formato(x))
```

## 5. Menú interactivo



```python
def menu():
    while True:
        print("""
====== REGISTRO DE ESTUDIANTES (BST) ======
1. Registrar estudiante (matrícula aleatoria)
2. Consultar por matrícula
3. Eliminar por matrícula
4. Listar en orden ascendente (inorden)
5. Recorridos preorden y postorden
6. Total de estudiantes
7. Altura del árbol
0. Salir""")
        op = input("Opción: ").strip()

        if op == "1":
            nombre = input("Nombre: ").strip()
            carrera = input("Carrera: ").strip()
            m = matricula_aleatoria(arbol)
            arbol.insertar(m, nombre, carrera)
            print(f">> Registrado con matrícula {m:06d}")

        elif op == "2":
            n = arbol.buscar(int(input("Matrícula: ")))
            print(">>", formato(n) if n else "No existe")

        elif op == "3":
            m = int(input("Matrícula: "))
            if arbol.buscar(m):
                arbol.eliminar(m)
                print(f">> Eliminado {m:06d}")
            else:
                print(">> No existe")

        elif op == "4":
            for x in arbol.inorden(arbol.root):
                print("  ", formato(x))

        elif op == "5":
            print("Preorden: ", [f"{x.matricula:06d}" for x in arbol.preorden(arbol.root)])
            print("Postorden:", [f"{x.matricula:06d}" for x in arbol.postorden(arbol.root)])

        elif op == "6":
            print(">> Total:", arbol.contar(arbol.root))

        elif op == "7":
            print(">> Altura:", arbol.altura(arbol.root))

        elif op == "0":
            print("Fin.")
            break

        else:
            print(">> Opción no válida")


menu()
```

## 6. Ejemplos de ejecución



### Caso 1 — Recorridos y estado del árbol
**Entrada:** los 7 estudiantes de ejemplo. **Salida:** recorridos, total y altura.

```python
print("Total:", arbol.contar(arbol.root), "| Altura:", arbol.altura(arbol.root))
print()
print("INORDEN (ascendente):")
for x in arbol.inorden(arbol.root):
    print("  ", formato(x))
print()
print("PREORDEN: ", [f"{x.matricula:06d}" for x in arbol.preorden(arbol.root)])
print("POSTORDEN:", [f"{x.matricula:06d}" for x in arbol.postorden(arbol.root)])
```

### Caso 2 — Alta con matrícula aleatoria y consulta
**Entrada:** se registra un estudiante nuevo y se consulta. **Salida:** matrícula generada y datos.

```python
random.seed(42)

m = matricula_aleatoria(arbol)
arbol.insertar(m, "Diego Ramírez", "Biología")
print(f"Registrado con matrícula {m:06d}")

print("Consulta:", formato(arbol.buscar(m)))
print("Consulta inexistente (000001):", arbol.buscar(1))
print("Total tras el alta:", arbol.contar(arbol.root))
```

### Caso 3 — Eliminación de un nodo con dos hijos
**Entrada:** se elimina la matrícula `101008`. **Salida:** el árbol se reorganiza y el inorden sigue ordenado.

```python
print("Antes: ", [f"{x.matricula:06d}" for x in arbol.inorden(arbol.root)])
print("  Altura:", arbol.altura(arbol.root), "| Total:", arbol.contar(arbol.root))

arbol.eliminar(101008)

print("Después:", [f"{x.matricula:06d}" for x in arbol.inorden(arbol.root)])
print("  Altura:", arbol.altura(arbol.root), "| Total:", arbol.contar(arbol.root))
```
